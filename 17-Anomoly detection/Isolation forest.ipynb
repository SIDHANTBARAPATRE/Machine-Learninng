```python
import pandas as pd
df = pd.read_csv("healthcare.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df.iloc[:,0], df.iloc[:,1], )
```

```python
from sklearn.ensemble import IsolationForest
```

```python
clf = IsolationForest(contamination=0.2) ## contamination is threshold.
clf.fit(df)
prediction = clf.predict(df)
```

```python
prediction
```

```python
import numpy as np
index = np.where(prediction < 0) ## this gives the index where the predictions are less than 0.
index
```
